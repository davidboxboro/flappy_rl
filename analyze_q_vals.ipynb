```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import dill
```

```python
with open('model.pkl', 'rb') as f:
    d = dill.loads(f.read())
```

```python
arr = [[[] for i in range(51)] for j in range(40)]
for k, v in d.items():
    if k[3] == 1:
        arr[(k[1] + 400) // 40][k[0] // 40].append(v)
arr = [[np.mean(f) for f in e] for e in arr]
sns.heatmap(arr)
```

```python
arr = [[[] for i in range(51)] for j in range(40)]
for k, v in d.items():
    if k[3] == 0:
        arr[(k[1] + 400) // 40][k[0] // 40].append(v)
arr = [[np.mean(f) for f in e] for e in arr]
sns.heatmap(arr)
```
